```python
import qrcode

# Your live GitHub Pages link
link = "https://sl1nk.com/umfr2z0"

# Initialize the QR Code object to control size
qr = qrcode.QRCode(
    version=1, # Controls the overall size of the QR code grid (1 is the smallest)
    error_correction=qrcode.constants.ERROR_CORRECT_L,
    box_size=5,  # Reduces the size of each individual pixel box (default is usually 10)
    border=4,    # Keeps a standard border thickness
)

# Add your link data
qr.add_data(link)
qr.make(fit=True)

# Generate the image, applying the red color
# fill_color dictates the QR code itself, back_color is the background
qr_img = qr.make_image(fill_color="red", back_color="white")

# Save the customized image locally
qr_img.save("CogSci_Cortex_QR_Small_Red.png")
print("Small, red QR code successfully saved!")
```

```python
import qrcode

# Your live GitHub Pages link
link = "https://rj041100.github.io/CogSci_Cortex/"

# Initialize the QR Code object optimized for small spaces
qr = qrcode.QRCode(
    version=1, 
    # Maximize error correction so it scans even if it's tiny or blurry
    error_correction=qrcode.constants.ERROR_CORRECT_H, 
    box_size=5,  # Keep the physical size small
    border=4,    # DO NOT shrink the border (explained below)
)

# Add your link data
qr.add_data(link)
qr.make(fit=True)

# Generate a high-contrast image (Black on White is always best for tiny codes)
qr_img = qr.make_image(fill_color="black", back_color="white")

# Save the robust image
qr_img.save("CogSci_Cortex_QR_Micro.png")
print("High-durability micro QR code successfully saved!")
```

```python
import qrcode

# Your live GitHub Pages link
link = "https://rj041100.github.io/CogSci_Cortex/"

# Initialize the QR Code object with zero border
qr = qrcode.QRCode(
    version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=5,
    border=0,    # <-- Set this to 0 to remove the margin completely
)

# Add your link data
qr.add_data(link)
qr.make(fit=True)

# Generate a high-contrast image
qr_img = qr.make_image(fill_color="black", back_color="white")

# Save the zero-margin image
qr_img.save("CogSci_Cortex_QR_NoMargin.png")
print("Zero-margin QR code successfully saved!")
```
